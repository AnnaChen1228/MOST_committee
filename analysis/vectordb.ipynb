```python
%cd /Users/anna_chen/Desktop/Recommand-System/MOST_committee
import csv
project_data_file = 'data/research_proj/115計算機學門審查/authors_projects.csv'
authors_dicts = csv.reader(open(project_data_file, encoding='utf-8'))
```

```python
import pandas as pd
from langchain_community.embeddings import HuggingFaceEmbeddings
from langchain_community.vectorstores import FAISS
from langchain.schema import Document

# 初始化 BGE 嵌入模型
embeddings = HuggingFaceEmbeddings(
    model_name="BAAI/bge-large-zh-v1.5",
    model_kwargs={'device': 'cpu'},
    encode_kwargs={'normalize_embeddings': True}
)

# 創建合併後的文檔
title_documents = []
keyword_documents = []

# 遍歷 authors_dicts 提取數據並合併
for author, data in authors_dicts.items():
    # 合併該作者的所有標題
    all_titles = "\n".join([title for title in data['title'] if title and isinstance(title, str)])
    if all_titles:  # 確保不是空字符串
        title_doc = Document(
            page_content=all_titles,
            metadata={"author": author, "type": "title"}
        )
        title_documents.append(title_doc)
    
    # 合併該作者的所有關鍵字
    all_keywords = "\n".join([keyword for keyword in data['keywords'] if keyword and isinstance(keyword, str)])
    if all_keywords:  # 確保不是空字符串
        keyword_doc = Document(
            page_content=all_keywords,
            metadata={"author": author, "type": "keyword"}
        )
        keyword_documents.append(keyword_doc)

# 創建標題的向量資料庫
print("正在創建標題向量資料庫...")
title_vectorstore = FAISS.from_documents(title_documents, embeddings)
print(f"標題向量資料庫創建完成，包含 {len(title_documents)} 個文檔")

# 創建關鍵字的向量資料庫
print("正在創建關鍵字向量資料庫...")
keyword_vectorstore = FAISS.from_documents(keyword_documents, embeddings)
print(f"關鍵字向量資料庫創建完成，包含 {len(keyword_documents)} 個文檔")

# 保存向量資料庫到本地
title_vectorstore.save_local("database/title_vectorstore_bge_merged")
keyword_vectorstore.save_local("database/keyword_vectorstore_bge_merged")
```
