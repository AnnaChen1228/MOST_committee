```python
import matplotlib.pyplot as plt

def plot_weight_pie(weights):
    # --- Mac 中文設定 ---
    plt.rcParams['font.sans-serif'] = ['Arial Unicode MS']
    plt.rcParams['axes.unicode_minus'] = False

    # 1. 準備資料與中文標籤對照
    label_mapping = {
        'title': '標題 (Title)',
        'keyword': '關鍵字 (Keyword)',
        'application_directions': '應用方向 (Application)',
        'problems_to_solve': '解決問題 (Problems)',
        'goals_to_achieve': '預期目標 (Goals)',
        'methods_to_solve': '解決方法 (Methods)'
    }
    
    labels = [label_mapping.get(k, k) for k in weights.keys()]
    sizes = list(weights.values())
    
    # 2. 設定配色方案 (使用柔和色系)
    colors = plt.get_cmap('Pastel1').colors
    
    # 3. 繪圖
    plt.figure(figsize=(10, 7))
    
    # wedgeprops 可以做出中間空心的「甜甜圈圖」，看起來更現代
    patches, texts, autotexts = plt.pie(
        sizes, 
        labels=labels, 
        autopct='%1.1f%%', 
        startangle=140, 
        colors=colors,
        pctdistance=0.85,
        # explode=[0.05 if s == max(sizes) else 0 for s in sizes] # 將最大佔比的區塊稍微突顯
    )

    # 4. 美化文字
    plt.setp(autotexts, size=20, weight="bold")
    plt.setp(texts, size=20)

    # 5. 加入中間的圓圈，變成甜甜圈圖 (選配)
    # centre_circle = plt.Circle((0,0), 0.70, fc='white')
    fig = plt.gcf()
    # fig.gca().add_artist(centre_circle)

    plt.title('各項指標權重分配圖', fontsize=20, pad=15)
    plt.axis('equal')  # 確保圓餅圖是圓的
    
    plt.tight_layout()
    plt.show()

# 執行
weights = {
    'title': 1,
    'keyword': 1,
    'application_directions': 3,
    'problems_to_solve': 3,
    'goals_to_achieve': 1,
    'methods_to_solve': 1
}
plot_weight_pie(weights)
```

```python
import matplotlib.pyplot as plt

def plot_weight_line(weights):
    # --- Mac 中文設定 ---
    plt.rcParams['font.sans-serif'] = ['Arial Unicode MS']
    plt.rcParams['axes.unicode_minus'] = False

    # 1. 準備資料與中文標籤對照 (修正了 keywords 的 s 匹配問題)
    label_mapping = {
        'title': '標題\n(Title)',
        'keywords': '關鍵字\n(Keywords)',
        'application_directions': '應用方向\n(Application)',
        'problems_to_solve': '解決問題\n(Problems)',
        'goals_to_achieve': '預期目標\n(Goals)',
        'methods_to_solve': '解決方法\n(Methods)'
    }
    
    # 提取標籤與數值
    labels = [label_mapping.get(k, k) for k in weights.keys()]
    values = list(weights.values())
    
    # 2. 繪圖設定
    plt.figure(figsize=(6, 6))
    
    # 繪製折線：使用青色 (teal)，加入圓點標記 (marker='o')
    plt.plot(labels, values, marker='o', linestyle='-', color='#008080', 
             linewidth=3, markersize=10, label='平均分數')
    
    # 填滿折線下方的區域 (讓圖表更有份量感)
    plt.fill_between(labels, values, color='#008080', alpha=0.1)

    # 3. 在每個點上方顯示具體數值
    for i, val in enumerate(values):
        plt.text(i, val + 0.01, f'{val:.3f}', ha='center', va='bottom', 
                 fontsize=12, fontweight='bold', color='#004d4d')

    # 4. 圖表美化
    plt.title('各項指標平均分數趨勢圖', fontsize=18, pad=20)
    plt.xlabel('評估指標', fontsize=14)
    plt.ylabel('平均分數', fontsize=14)
    
    # 設定 Y 軸範圍，讓波動更明顯 (可根據資料調整)
    plt.ylim(min(values) - 0.05, max(values) + 0.1)
    
    # 加入水平網格線，方便對齊數值
    plt.grid(axis='y', linestyle='--', alpha=0.6)
    
    plt.tight_layout()
    plt.show()

# --- 執行資料 ---
weights = {
    'title': 0.389, 
    'keywords': 0.485, 
    'application_directions': 0.475, 
    'problems_to_solve': 0.487, 
    'goals_to_achieve': 0.534, 
    'methods_to_solve': 0.587
}

plot_weight_line(weights)
```
